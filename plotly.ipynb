```python
import plotly.figure_factory as ff
```

```python
ff.create_2d_density( df["sepal_width"], df["sepal_length"], colorscale="Viridis")
```

```python
ff.create_annotated_heatmap(
    z=df.pivot_table(index="species", columns="sepal_width", values="sepal_length").values,
    x=df.pivot_table(index="species", columns="sepal_width", values="sepal_length").columns.tolist(),
    y=df.pivot_table(index="species", columns="sepal_width", values="sepal_length").index.tolist(),
    colorscale="Viridis",)
```

```python
ff.create_dendrogram(
    df[["sepal_width", "sepal_length", "petal_width", "petal_length"]].values,
    orientation="left",
    labels=df["species"].tolist())
```

```python
ff.create_distplot(
    [df[df["species"] == species]["sepal_length"] for species in df["species"].unique()],
    group_labels=df["species"].unique().tolist(),
    colors=["red", "green", "blue"],
)
```

```python
ff.create_gantt(
    df[["species", "sepal_width", "sepal_length"]].rename(columns={"species": "Task", "sepal_width": "Start", "sepal_length": "Finish"}),
    index_col="Task",
    show_colorbar=True,
    group_tasks=True,
)
```

```python
ff.create_facet_grid(
    df,
    x="sepal_width",
    y="sepal_length",
    facet_col="species")
```

```python
ff.create_quiver(
    df["sepal_width"],
    df["sepal_length"],
    df["petal_width"],
    df["petal_length"],
    scale=0.1)
```
