# 🏷️ Part-of-Speech (POS) Tagging & Named Entity Recognition (NER)

## Objectives
In this notebook, you will learn how to:
- Use **NLTK** for Part-of-Speech (POS) Tagging on customer feedback.
- Use **spaCy** for Named Entity Recognition (NER) to extract entities (e.g., people, organizations, dates, locations, products) from unstructured customer feedback.
- Process, tabulate, and visualize the linguistic attributes of unstructured text.

```python
import nltk
import spacy
import pandas as pd
from spacy import displacy

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('averaged_perceptron_tagger')

nlp = spacy.load('en_core_web_sm')
```

```python
feedback_sentences = [
    "I bought an iPhone 15 from the Apple Store in New York last Friday. John Doe was incredibly helpful!",
    "The customer support at Microsoft was exceptional. I spoke with Sarah on Tuesday, and she resolved my Azure subscription issue quickly.",
    "We visited the Starbucks in Seattle and ordered three lattes for $15 total.",
    "The package from Amazon arrived 2 days late in San Francisco, but the refund of $10 was processed immediately.",
    "Tesla's Model 3 is amazing, and I am planning to take it on a road trip to the Grand Canyon next June."
]

for idx, sentence in enumerate(feedback_sentences, 1):
    print(f"{idx}. {sentence}")
```

POS tagging labels each word with its corresponding part of speech (noun, verb, adjective, adverb, etc.) based on both its definition and its context.

```python
def perform_nltk_pos_tagging(text):
    tokens = nltk.word_tokenize(text)
    # Perform POS Tagging
    pos_tags = nltk.pos_tag(tokens)
    return pos_tags

print("=== Running POS Tagging on Sentence 1 ===")
sample_tags = perform_nltk_pos_tagging(feedback_sentences[0])

# Convert to a structured DataFrame for clean display
df_pos = pd.DataFrame(sample_tags, columns=['Token', 'POS Tag'])
df_pos
```

```python
all_pos_data = []
for idx, sentence in enumerate(feedback_sentences, 1):
    pos_tags = perform_nltk_pos_tagging(sentence)
    for token, tag in pos_tags:
        all_pos_data.append({
            "Sentence ID": idx,
            "Token": token,
            "POS Tag": tag
        })

df_all_pos = pd.DataFrame(all_pos_data)
print(f"Total tokens tagged across all sentences: {len(df_all_pos)}")
df_all_pos.head(15)
```

Named Entity Recognition (NER) identifies and categorizes key entities in unstructured text into predefined categories like names of people, organizations, locations, monetary values, percentages, and dates.

```python
def perform_spacy_ner(text):
    doc = nlp(text)
    entities = []
    for ent in doc.ents:
        entities.append({
            "Entity Text": ent.text,
            "Label (Short)": ent.label_,
            "Label Explanation": spacy.explain(ent.label_),
            "Start Char": ent.start_char,
            "End Char": ent.end_char
        })
    return entities

print("=== Extracting Named Entities from Customer Feedback ===")
all_entities = []
for idx, sentence in enumerate(feedback_sentences, 1):
    ents = perform_spacy_ner(sentence)
    for ent in ents:
        ent["Sentence ID"] = idx
        all_entities.append(ent)

df_entities = pd.DataFrame(all_entities)
df_entities = df_entities[["Sentence ID", "Entity Text", "Label (Short)", "Label Explanation", "Start Char", "End Char"]]
df_entities
```

Visualizing Entities with displaCy


```python
for idx, sentence in enumerate(feedback_sentences, 1):
    print(f"\n--- Sentence {idx} NER Visualized ---")
    doc = nlp(sentence)
    displacy.render(doc, style="ent", jupyter=True)
```

## 📊 5. Summary and Comparison

| Task | Tool | Level of Analysis | Primary Output |
| :--- | :--- | :--- | :--- |
| **POS Tagging** | **NLTK** | Word/Token Level | Grammatical categories (Noun, Verb, Adjective, etc.) for *every* token. |
| **Named Entity Recognition (NER)** | **spaCy** | Phrase/Entity Level | Real-world semantic classes (PERSON, ORG, GPE, DATE, etc.) for *specific* spans of text. |
